# EDA: CIC IoMT WiFi MQTT Dataset (Parquet)

This notebook performs exploratory data analysis on:
- `data/CIC_IoMT_2024_WiFi_MQTT_all.parquet`

What is included:
1. Load and basic sanity checks
2. Schema and memory profile
3. Missing values
4. Label distribution and imbalance
5. Numeric feature summaries
6. Correlation analysis (sampled)
7. Outlier checks and distribution plots
8. Save a concise EDA summary

```python
from pathlib import Path
import pandas as pd

data_path = Path("../../data/CIC_IoMT_2024_WiFi_MQTT_all.parquet").resolve()

print("Exists:", data_path.exists())
print("Path:", data_path)

df = pd.read_parquet(data_path)
print("\nShape:", df.shape)
print("\nColumns:", len(df.columns))
display(df.head(5))
```

```python
df.columns.tolist()
for c in df.columns:
    print(c)
```

```python


# show full rows without truncation
pd.set_option("display.max_columns", None)
pd.set_option("display.width", None)
pd.set_option("display.max_colwidth", None)

# 5 full rows where:
# - label contains "MQTT"
# - label is NOT benign
attack_mqtt_5 = df[
    df["label"].str.contains("MQTT", case=False, na=False) &
    ~df["label"].str.contains("Benign", case=False, na=False)
].head(5000)

print(attack_mqtt_5.to_string(index=False))
```
